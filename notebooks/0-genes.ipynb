```python
get_codons = lambda nt_seq : [nt_seq[i:i + 3] for i in range(0, len(nt_seq), 3)]


def orfm_parser(record, start_codons:list=['ATG', 'GTG', 'TTG'], contigs:dict=contigs):
    '''

    https://github.com/wwood/OrfM
    '''
    start, frame, n = record.id.split('_')[-3:] # OrfM gene IDs are of the form {sequence_name}_{start}_{frame}_{gene_number}.
    contig = contigs[get_genome_id(record.id)]

    start = int(start) if (int(frame) <= 3) else (int(start) - 3) # The start coordinate is one-indexed (inclusive).
    stop = (start - 1) + ((len(str(record.seq)) + 1) * 3) # The stop coordinate, including the stop codon. This is also one-indexed (inclusive).
    # stop = stop + 3 if (int(frame) <= 3) else stop # The start coordinate is one-indexed (inclusive).

    # When OrfM identifies a start or stop codon near a contig border, the resulting ORF may be incomplete. This means that, depending on 
    # which gene end is partial, either:
    #   (1) The ORF will not have a valid stop codon, or
    #   (2) Computing the stop coordinate as start + ((len(seq) + 1) * 3) may result stop > len(contig), and a nt_seq indivisible by 3.

    nt_seq = get_reverse_complement(contig[start - 1:stop]) if (int(frame) > 3) else contig[start - 1:stop]
    codons = get_codons(nt_seq)
    
    if ((len(nt_seq) % 3) != 0) or (stop > len(contig)) or ((start - 1) < 0):
        # assert stop > len(contig), 'orfm_parse_entry: Expected an invalid stop coordinate if the nucleotide sequence is indivisible by 3'
        return {'partial':True, 'gene_id':f'orfm.{get_genome_id(record.id)}.1_{n}'}

    assert (stop - start) % 3 == 2, f'prodigal_parser: Expected the stop and start coordinates to be one-indexed and inclusive.' 
    assert (len(nt_seq) % 3) == 0, f'orfm_parse_entry: Expected the nucleotide sequence to be divisible by 3.' 
    assert (len(codons) - 1) == len(str(record.seq)), f'orfm_parse_entry: Expected the number of codons to be the same length as the translated sequence.' 

    data = dict()
    data['genome_size'] = len(contig)
    data['start'] = start 
    data['stop'] = stop
    data['gene_id'] = f'orfm.{get_genome_id(record.id)}.1_{n}'
    data['seq'] = str(record.seq)
    data['original_gene_id'] = record.id
    data['genome_id'] = get_genome_id(record.id)
    data['nt_seq'] = nt_seq
    data['frame'] = frame 
    data['strand'] = '+' if (int(frame) <= 3) else '-'
    data['has_start_codon'] = np.any(np.isin(codons, start_codons))
    if data['has_start_codon']:
        data['start_codon_idx'] = np.where(np.isin(codons, start_codons))[0][0]
        data['start_codon'] = codons[data['start_codon_idx']]
    data['stop_codon'] = codons[-1]
    data['partial'] = False #(data['stop_codon'] not in stop_codons)
    return data



def prodigal_parser(record, contigs:dict=contigs):
    ''''''
    
    header_pattern = r'# (?P<start>\d+) # (?P<stop>\d+) # (?P<strand>[\d\-]+) # ID\=\d+_(?P<gene_number>\d+)'

    start, stop, strand, n = re.search(header_pattern, record.description).groups()
    start, stop = int(start), int(stop)

    data = dict()
    data['stop'] = stop
    data['start'] = start
    data['partial'] = re.search(r'partial=(\d+)', record.description).group(1) != '00' # True if the partial string is not 00.
    data['strand'] = '+' if (strand == '1') else '-'
    data['start_codon'] = re.search(r'start_type=([^;]+)', record.description).group(1)
    data['seq'] = str(record.seq).replace('*', '')
    data['original_gene_id'] = record.id
    data['frame'] = 0
    data['gene_id'] = f'prodigal.{get_genome_id(record.id)}.1_{n}'
    data['genome_id'] = get_genome_id(record.id)

    contig = contigs[data['genome_id']]
    nt_seq = get_reverse_complement(contig[start - 1:stop]) if (strand == '-1') else contig[start - 1:stop]
    codons = get_codons(nt_seq)

    assert (codons[0] == data['start_codon']) or data['partial'], f'prodigal_parser: The first codon {codons[0]} does not match the expected start codon {data['start_codon']}.' 
    assert (stop - start) % 3 == 2, f'prodigal_parser: Expected the stop and start coordinates to be one-indexed and inclusive.' 
    assert (len(nt_seq) % 3) == 0, f'prodigal_parser: Expected the nucleotide sequence to be divisible by 3.' 
    assert (len(codons) - 1) == len(data['seq']), f'prodigal_parser: Expected the number of codons to be the same length as the translated sequence.' 

    data['nt_seq'] = nt_seq 
    data['stop_codon'] = codons[-1]
    data['genome_size'] = len(contig)

    return data
```

```python
def get_prodigal_gene_id(row, genes_df=None):
    '''For an OrfM prediction, get the Prodigal-predicted gene which corresponds to it. 
    
    :param row
    :param genes_df
    :returns: 
    '''
    filter_ = (row.genome_id == genes_df.genome_id)
    filter_ = filter_ & (genes_df.strand == row.strand) # Make sure ORFs are predicted on the same strand. 
    filter_ = filter_ & ((genes_df.stop == row.stop) | (genes_df.start == row.start))

    assert filter_.sum() <= 1, f'get_prodigal_gene_id: Expecting only one corresponding Prodigal gene, but got {filter_.sum()}.'
    return None if (filter_.sum() == 0) else genes_df[filter_].gene_id.values[0]
```

```python
# https://journals.plos.org/ploscompbiol/article?id=10.1371/journal.pcbi.1008439

def get_gc_skew_info(nt_seq, window_size=2000, step_size=500, min_window_size=300, use_reverse_complement:bool=False):
    '''Preliminary inspection of the GC skew graphs indicates that all Betazoid genomes, like Borgs, have two uneven replichores.
    The current theory for Borg replication is that it begins at either terminus (the ITRs) and proceeds inwards; this would mean that the
    interior GC skew transition point is a ter. All genes on the forward strand prior to ter would be on the leading strand of one replichore, 
    and all genes on the reverse strand after ter would be on the leading strand of the other replichore.'''
    nt_seq = get_reverse_complement(nt_seq) if use_reverse_complement else nt_seq

    info = dict()

    genome_size = len(nt_seq)

    # Because one of the replichores are so short, a window size of 5000 does not pick up the sign change in GC skew. However, if the window size is
    # decreased, GC skew becomes too noisy. To fix this, just adjusted the window sizes to be shorter at either ITR.

    n = ((window_size - min_window_size) // step_size) + 1 # The number of shorter windows to add at the left end. 
    # print(f'get_gc_skew_info: Splitting the {window_size} bp region at the left terminus into {n} windows.')

    window_starts = [0] * n 
    window_stops = list(range(min_window_size, window_size + 1, step_size))
    assert len(window_starts) == len(window_stops), f'get_gc_skew_info: There are {len(window_starts)} starts and {len(window_stops)} stops.'
    window_starts = np.array(window_starts + list(range(window_size, genome_size, step_size)))
    window_stops = np.array(window_stops + [min(genome_size, i + window_size) for i in range(window_size, genome_size, step_size)])
    assert len(window_starts) == len(window_stops), f'get_gc_skew_info: There are {len(window_starts)} starts and {len(window_stops)} stops.'
    
    windows = np.array([nt_seq[start:stop] for start, stop in zip(window_starts, window_stops)])

    mask = np.array([len(window) for window in windows]) > min_window_size
    window_starts = np.array(window_starts[mask]).copy()
    window_stops = np.array(window_stops[mask]).copy()
    windows = windows[mask].copy()

    # print(f'get_gc_skew_info: Discarding {(~mask).sum()} windows shorter than {min_window_size} bp')
    # print(f'get_gc_skew_info: {(window_starts == 0).sum()} windows starting at position 0.')
    # print(f'get_gc_skew_info: {(window_stops == genome_size).sum()} windows ending at position {genome_size}.')

    info['window_midpoints'] = (window_stops + window_starts) / 2
    info['window_midpoints_relative'] = info['window_midpoints'] / genome_size
    info['window_starts'] = window_starts
    info['window_stops'] = window_stops
    info['window_starts_relative'] = window_starts / genome_size

    _get_gc_skew = lambda window : (window.count('G') - window.count('C')) / (window.count('G') + window.count('C'))
    info['gc_skews'] = np.array([_get_gc_skew(window) for window in windows]) # One GC skew per window. 
    info['gc_skews_cumulative'] = np.cumsum(info['gc_skews'])

    info['gc_skew_transition_idxs'] = np.where(np.diff(np.sign(info['gc_skews'])) != 0)[0]
    return info

# Unsure why the correct ter GC skew flip is not registering for bz_0; possible that the small replichore is too short.
# Inspection of the gene orientations indicates that it should be reverse-complemented for ter to be on the correct side, so did this manually. 
# Ultimately decided to locate all GC skew transitions manually to be safe. 

# The template strand should be G-depleted, so GC skew should be decreasing (more negative) over the replicon, assuming replication proceeds 
# from the termini. 

genome_ids = list(contig_ids.keys())
fig, ax_1 = plt.subplots(figsize=(5, 4))
ax_2 = ax_1.twinx()

for genome_id in genome_ids:
    if genome_id == 'bz_12':
        continue
    if genome_id != 'bz_0':
        continue
    genome = get_reverse_complement(GENOME_CONTIGS[genome_id])
    genome = borg_genome

    info = get_gc_skew_info(genome, use_reverse_complement=(genome_id in reverse_complement_genome_ids), window_size=100, min_window_size=50)
    ax_1.plot(info['window_midpoints'], info['gc_skews_cumulative'], color=GENOME_ID_COLORS[genome_id], label=genome_id)
    ax_2.plot(info['window_midpoints'], info['gc_skews'], color=GENOME_ID_COLORS[genome_id], label=genome_id, ls='--', lw=0.5)

ax_1.set_ylabel('cumulative GC skew')
ax_2.set_ylabel('GC skew')

ax_1.set_xlabel('position')
# ax.legend()
fig.tight_layout()
plt.show() 
    
    
```
